import json
import random
from collections import defaultdict
from PIL import Image
import matplotlib.pyplot as plt

# notebook is inside notebooks/, so go one level up
ROOT = ".."

# load COCO val annotations
with open(f"{ROOT}/annotations/instances_val2017.json") as f:
    anns = json.load(f)

with open(f"{ROOT}/annotations/captions_val2017.json") as f:
    caps = json.load(f)

# pick 500 images for testing (same 500 every run)
random.seed(42)
test_images = random.sample(anns["images"], 500)
test_files = [img["file_name"] for img in test_images]
test_ids = {img["id"] for img in test_images}

# captions for those 500 images
captions_by_image = defaultdict(list)
for ann in caps["annotations"]:
    if ann["image_id"] in test_ids:
        captions_by_image[ann["image_id"]].append(ann["caption"])

print(f"Test set size: {len(test_files)}")
print("First 5 files:", test_files[:5])

```python
# show one image from the 500-image test set
sample = test_images[0]
img = Image.open(f"{ROOT}/val2017/{sample['file_name']}")

plt.imshow(img)
plt.axis("off")
plt.title(sample["file_name"])
plt.show()

print("Captions:")
for c in captions_by_image[sample["id"]]:
    print("-", c)
```
